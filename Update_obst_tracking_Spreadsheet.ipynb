```python
import pandas as pd
```

```python
observed = pd.read_csv('targets_already_scheduled.csv')
```

```python
# Using the footprint that exclude the targets to block
# SAVING TIME for some pulsar search survey things 
footprint = pd.read_csv('MMGPS-SBAND_pointing_grid_final_blocked.csv')
full_footprint = pd.read_csv('MMGPS-SBAND_pointing_grid_final.csv')
```

```python
# Find the blocked pointings (targets in a full footprint, not in a blocked footprint)
blocked = full_footprint[~full_footprint['Pointing'].isin(footprint['Pointing'])]

# Find the observed Targets that are in full footprint 
observed = full_footprint[full_footprint['Pointing'].isin(observed['Pointing'])]

# Find outstanding targets (Not observed) 
# outstanding = full_footprint[~full_footprint['Pointing'].isin(observed['Pointing'])]
```

```python
with pd.ExcelWriter('Obs_tracking.xlsx', engine='xlsxwriter') as writer:
    # Merge Full footprint and observed targets 
    df_merged = pd.merge(full_footprint.drop(columns=['Gb','RA', 'Dec']), observed.drop(columns=['Gb','RA', 'Dec']), on='Pointing', how='left')

    # Drop the unnecessary  Glx column and replace the observed Glx with either False if NaN 
    # or True if not (To show that the observation is observed or not); 
    # Re-label the column keys accordingly 
    df = df_merged.drop(columns=['Gl_x'])
    df['Gl_y'] = ~df['Gl_y'].isna()
    new_df = df.rename(columns={'Gl_y': 'Observed/Scheduled?'})
    new_df.to_excel(writer, sheet_name='Sheet1', index=False)
new_df
```
