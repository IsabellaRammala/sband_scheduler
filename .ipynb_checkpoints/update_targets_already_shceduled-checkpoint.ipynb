```python
import pandas as pd
import astropy.time
import astropy.coordinates
```

```python
mmgps_database_observed = pd.read_csv('mmgps_sband_observed.csv')
mmgps_database_observed
```

```python
# Convert utc_start time to lst for meerKAT @ location -30.71322, 21.44306 (Coordinates in decimal)
mkat_lat = -30.71322
mkat_lon = 21.44306
obs_time_utc = mmgps_database_observed['utc_start']

utc_time = astropy.time.Time(val=obs_time_utc.to_list(),
                                location=astropy.coordinates.EarthLocation.from_geodetic(lon=mkat_lon, lat=mkat_lat))
lst_time = utc_time.sidereal_time("apparent").to_string(sep=":")
lst_time
```

```python
#Pointing	RA	DEC	Obs_time_UTC	Obs_Time_LST
#(target str)   (ra hh:mm:ss.s)  (dec dd:mm:ss.s)   (utc yyyy-mm-dd hh:mm:ss)  (LST hh:mm:ss.ssssss)
new_df = mmgps_database_observed.drop(columns='pointing_id').loc[:,['target','ra','dec', 'utc_start']]
new_df_column_renamed = new_df.rename(columns={'target': 'Pointing','ra':'RA','dec':'DEC','utc_start':'Obs_time_UTC'})
new_df_column_renamed['Obs_Time_LST'] = lst_time
new_df_column_renamed.to_csv('targets_already_scheduled.csv',index=False)
```
